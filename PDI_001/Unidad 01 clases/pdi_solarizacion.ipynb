SOLARIZACION

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def solarizar (imagen, umbral):
    imagen_solarizada=np.where(imagen<umbral, imagen, 255-imagen)
    return imagen_solarizada

imagen = cv2.imread('imagentango.jpg')
img_gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
umbral=127

imagen_nueva = solarizar(imagen, umbral)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
plt.subplot(1,2,2)
plt.imshow(imagen_nueva,cmap='gray')
plt.show()
```

SOLARIZACION

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def posterizar (imagen, nivel):
    factor= 255/nivel
    image_post= np.round(imagen/(factor)*factor)
    return image_post.astype(np.uint8)

imagen = cv2.imread('imagentango.jpg')
img_gris = cv2.cvtColor (imagen, cv2.COLOR_BGR2GRAY)
nivel=4 #nivel de intensidad
nueva= posterizar(imagen, nivel)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))
plt.subplot(1,2,2)
plt.imshow(nueva,cmap='gray')
plt.show()
```

FALSO COLOR

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def falso_color(imagen_gray):
    escalar_imagen= cv2.normalize(imagen_gray, None, 0, 255, cv2.NORM_MINMAX)
    
    aplio_color= cv2.applyColorMap(escalar_imagen, cv2.COLORMAP_DEEPGREEN)
    return aplio_color

image_gray = cv2.imread('imagentango.jpg')
img_falso = falso_color(image_gray)

plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(image_gray, cv2.COLOR_BGR2RGB))
plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_falso, cv2.COLOR_BGR2RGB))
plt.show()
```
