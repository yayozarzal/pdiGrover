CREACION DE UNA IMAGEN

```python
import matplotlib.pyplot as plt
import numpy as np

# Configuración del tamaño de la figura
plt.rcParams['figure.figsize'] = [10, 5]

# Definir el tamaño de la imagen
tamano_imagen = 100

# Crear matrices para cada canal de color
R = np.ones((tamano_imagen, tamano_imagen)) * 255  # Rojo intenso
G = np.ones((tamano_imagen, tamano_imagen)) * 128  # Verde medio
B = np.ones((tamano_imagen, tamano_imagen)) * 10   # Azul tenue

# Combinar los canales en una imagen RGB
imagen = np.dstack((R, G, B))  # Se combinan los 3 canales

# Convertir la imagen a tipo entero (int) para evitar errores
imagen = np.array(imagen, dtype=np.uint8)

# Mostrar la imagen
plt.figure()
plt.imshow(imagen)
plt.axis("on")  # Ocultar ejes
plt.show()
```

ESCALA DE GRISES Y PROFUNDIDA DE COLOR

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np


imagen = cv2.imread('imagentango.jpg', cv2.IMREAD_GRAYSCALE)

# Verificar si la imagen se cargó correctamente
if imagen is None:
    print("Error: No se pudo cargar la imagen. Verifica el nombre del archivo.")
else:
    # Obtener la profundidad de color original
    img_inicial = imagen.dtype
    print('Profundidad de color inicial:', img_inicial)  # Debería ser uint8 (8 bits)

    
    img_16bit = (imagen.astype(np.uint16)) * 256  # Multiplicamos por 256 para expandir el rango 0-255 → 0-65535


    img_nueva = img_16bit.dtype
    print('Profundidad de color nueva:', img_nueva)  # Debería ser uint16 (16 bits)

    # imagen or y conberrida
    plt.figure(figsize=(10, 5))

    # Imagen original (8 bits)
    plt.subplot(121)
    plt.title("Imagen Original (8 bits)")
    plt.imshow(imagen, cmap='gray')
    plt.axis("off")

    # Imagen convertida a 16 bits (Matplotlib no soporta 16 bits, pero se muestra como 8 bits para visualización)
    plt.subplot(122)
    plt.title("Imagen Convertida (16 bits)")
    plt.imshow(img_16bit // 256, cmap='gray')  # Normalizamos para que sea visible en 8 bits
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # Guardar la imagen convertida a 16 bits en el ordenador
    save_path = "imagen_16bit.png"
    cv2.imwrite(save_path, img_16bit)
    print(f"Imagen en 16 bits guardada en: {save_path}")
```

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

# Crear la figura con 4 subgráficos en una fila
fig, axs = plt.subplots(1, 4, figsize=(25, 10))

# Cargar la imagen en BGR y convertir a RGB
img_bgr = cv2.imread('imagentango.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Convertir la imagen de RGB a HSV
hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)

# Mostrar la imagen original en RGB
axs[0].imshow(img_rgb)
axs[0].set_title('Imagen Original')  # Corrección aquí
axs[0].axis("off")

# Mostrar el canal H (Tono) en escala de grises
fig_img1 = axs[1].imshow(hsv[..., 0], cmap='gray', vmin=0, vmax=179)
fig.colorbar(fig_img1, ax=axs[1], shrink=0.5)
axs[1].set_title('Canal H (Tono)')  # Corrección aquí
axs[1].axis("off")

# Mostrar el canal S (Saturación)
fig_img2 = axs[2].imshow(hsv[..., 1], cmap='gray', vmin=0, vmax=255)
fig.colorbar(fig_img2, ax=axs[2], shrink=0.5)
axs[2].set_title('Canal S (Saturación)')
axs[2].axis("off")

# Mostrar el canal V (Brillo)
fig_img3 = axs[3].imshow(hsv[..., 2], cmap='gray', vmin=0, vmax=255)
fig.colorbar(fig_img3, ax=axs[3], shrink=0.5)
axs[3].set_title('Canal V (Brillo)')
axs[3].axis("off")

# Ajustar el layout y mostrar
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

Imagen = cv2.imread('imagentango.jpg')
img= cv2.cvtColor(Imagen, cv2.COLOR_BGR2RGB)
def RGB2HSV(img):
    img = img.astype(np.float32)
    img = img/255.
    new_img = np.zeros_like(img)

    maxChannel = np.argmax(img,axis=2)
    maxValue = np.amax(img,axis=2)
    minChannel = np.argmin(img,axis=2)
    minValue = np.amin(img,axis=2)
    #value 
    new_img[...,2] = maxValue

    #saturation
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if maxValue[i,j]==0:
                new_img[i,j,1] = 0
            elif maxValue[i,j] != 0:
                new_img[i,j,1] = (new_img[i,j,2]-minValue[i,j])/new_img[i,j,2]

    #hue 
    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            if sum(img[i,j,...])/3 == img[i,j,0]:
                new_img[i,j,0] = 0
            elif maxChannel[i,j] == 0: #R 
                new_img[i,j,0] = 60 * ((img[i,j,1]-img[i,j,2])/(new_img[i,j,2]-minValue[i,j]))
            elif maxChannel[i,j] == 1: #G
                new_img[i,j,0] = 120 + 60 * (img[i,j,2]-img[i,j,0])/(new_img[i,j,2]-minValue[i,j])
            elif maxChannel[i,j] == 2: #B
                new_img[i,j,0] = 240 + 60 * (img[i,j,0]-img[i,j,1])/(new_img[i,j,2]-minValue[i,j])

        bol_ind = new_img[...,0] < 0
        new_img[bol_ind,0]   =  new_img[bol_ind,0] + 360
    return new_img 

hsv_img =  RGB2HSV(img) 
fig, axs = plt.subplots(1,4,figsize=(25,10))
axs[0].imshow(img)
axs[0].set_title('Imagen')
fig_img = axs[1].imshow(hsv_img[...,0],cmap='gray',vmin=0,vmax=360)
fig.colorbar(fig_img, ax=axs[1],shrink=0.3)
axs[1].set_title('Canal Hue')
fig_img = axs[2].imshow(hsv_img[...,1],cmap='gray',vmin=0,vmax=1)
fig.colorbar(fig_img, ax=axs[2],shrink=0.3)
axs[2].set_title('Canal Saturación')
fig_img = axs[3].imshow(hsv_img[...,2],cmap='gray',vmin=0,vmax=1)
fig.colorbar(fig_img, ax=axs[3],shrink=0.3)
axs[3].set_title('Canal Value');
```

```python
def HSV2RGB(img):
    new_img = np.zeros_like(img)
    
    C = img[...,2]*img[...,1]
    X = C*(1-abs((img[...,0]/60)%2-1))
    m = img[...,2]-C

    for i in range(img.shape[0]):
        for j in range(img.shape[1]):
            h = img[i,j,0]
            if 0 <= h and h < 60:
                new_img[i,j] = C[i,j],X[i,j],0
            elif 60 <= h and h < 120:
                new_img[i,j] = X[i,j],C[i,j],0
            elif 120 <= h and h < 180:
                new_img[i,j] = 0,C[i,j],X[i,j]
            elif 180 <= h and h < 240:
                new_img[i,j] = 0,X[i,j],C[i,j]
            elif 240 <= h and h < 300:
                new_img[i,j] = X[i,j],0,C[i,j]
            elif 300 <= h and h < 360:
                new_img[i,j] = C[i,j],0,X[i,j]
    new_img[...,0] = (new_img[...,0]+m)*255
    new_img[...,1] = (new_img[...,1]+m)*255
    new_img[...,2] = (new_img[...,2]+m)*255
    new_img = new_img.astype(int)
    return new_img 
rgb_img =   HSV2RGB(hsv_img) 
fig, axs = plt.subplots(1,4,figsize=(25,10))

axs[0].imshow(img)
axs[0].set_title('Imagen')


fig_img = axs[1].imshow(rgb_img[...,0],cmap='gray',vmin=0,vmax=255)
fig.colorbar(fig_img, ax=axs[1],shrink=0.3)
axs[1].set_title('Canal Rojo')


fig_img = axs[2].imshow(rgb_img[...,1],cmap='gray',vmin=0,vmax=255)
fig.colorbar(fig_img, ax=axs[2],shrink=0.3)
axs[2].set_title('Canal Verde')


fig_img = axs[3].imshow(rgb_img[...,2],cmap='gray',vmin=0,vmax=255)
fig.colorbar(fig_img, ax=axs[3],shrink=0.3)
axs[3].set_title('Canal Azul');
```

umbral binario, umbral binario inverso

```python
import cv2
import matplotlib.pyplot as plt

# Cargar la imagen en escala de grises
imagen = cv2.imread('imagentango.jpg', 0)

# Aplicar umbralización binaria
_, umbralBinario = cv2.threshold(imagen, 127, 255, cv2.THRESH_BINARY)

# Aplicar umbralización binaria inversa
_, umbralBinarioInverso = cv2.threshold(imagen, 127, 255, cv2.THRESH_BINARY_INV)

# Mostrar las imágenes
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(imagen, cmap='gray')
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(umbralBinario, cmap='gray')
plt.title('Umbral Binario')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(umbralBinarioInverso, cmap='gray')
plt.title('Umbral Binario Inverso')
plt.axis('off')

plt.show()
```
