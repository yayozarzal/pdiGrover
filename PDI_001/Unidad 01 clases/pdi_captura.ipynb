```python
import cv2
import numpy as np
import matplotlib.pyplot as plt  # Ojo con 'matplotlib.pylab' vs. 'matplotlib.pyplot'

# Usa un raw string o corrige las barras
img_bgr = cv2.imread('imagentango.jpg')      # Lee la imagen en BGR (por defecto)

# Si img_bgr es None, revisa que la ruta sea correcta
if img_bgr is None:
    print("No se pudo leer la imagen. Verifica la ruta.")
else:
    # Convierte a RGB para mostrarla bien en plt
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    # Muestra la imagen
    plt.imshow(img_rgb)
    plt.axis("off")  # Quita los ejes
    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np


plt.rcParams ['figure.figsize']=[10,5]

# path = r"D:\UNIVALLE\SISTEMAS\5TO SEMESTRE\PROCESAMIENTO DIGITAL DE IMAGENES\TEMA 1\JPEG\imagentango.jpg"
img = cv2.imread('imagentango.jpg', cv2.IMREAD_COLOR)
img_rgb=cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

r=img[:,:,2]
g=img[:,:,1]
b=img[:,:,0]


red_imagen=np.zeros_like(img_rgb)
red_imagen[:,:,0]=r

green_imagen=np.zeros_like(img_rgb)
green_imagen[:,:,1]=g

blue_imagen=np.zeros_like(img_rgb)
blue_imagen[:,:,2]=b

plt.subplot(141)
plt.title("canal original")
plt.imshow(img_rgb)
plt.axis("off")

plt.subplot(142)
plt.title("canal rojo")
plt.imshow(red_imagen)
plt.axis("off")

plt.subplot(143)
plt.title("canal verde")
plt.imshow(green_imagen)
plt.axis("off")

plt.subplot(144)
plt.title("canal azul")
plt.imshow(blue_imagen)
plt.axis("off")


plt.show()
```
