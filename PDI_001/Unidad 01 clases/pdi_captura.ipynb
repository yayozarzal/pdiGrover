CREACION DE UNA IMAGEN

```python
import matplotlib.pyplot as plt
import numpy as np

# Configuración del tamaño de la figura
plt.rcParams['figure.figsize'] = [10, 5]

# Definir el tamaño de la imagen
tamano_imagen = 100

# Crear matrices para cada canal de color
R = np.ones((tamano_imagen, tamano_imagen)) * 255  # Rojo intenso
G = np.ones((tamano_imagen, tamano_imagen)) * 128  # Verde medio
B = np.ones((tamano_imagen, tamano_imagen)) * 10   # Azul tenue

# Combinar los canales en una imagen RGB
imagen = np.dstack((R, G, B))  # Se combinan los 3 canales

# Convertir la imagen a tipo entero (int) para evitar errores
imagen = np.array(imagen, dtype=np.uint8)

# Mostrar la imagen
plt.figure()
plt.imshow(imagen)
plt.axis("on")  # Ocultar ejes
plt.show()
```

como podria crear una imagen pero adicionando un pixel de otro color dentro de este 1

```python
import matplotlib.pyplot as plt
import numpy as np


tamano_imagen = 100


R = np.ones((tamano_imagen, tamano_imagen)) * 255  
G = np.ones((tamano_imagen, tamano_imagen)) * 128  
B = np.ones((tamano_imagen, tamano_imagen)) * 10   


imagen = np.dstack((R, G, B)).astype(np.uint8) 


imagen[50, 50] = [0, 255, 0]  

# Mostrar la imagen
plt.figure()
plt.imshow(imagen)
plt.axis("on") 
plt.show()
```

ESCALA DE GRISES Y PROFUNDIDA DE COLOR

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np


imagen = cv2.imread('imagentango.jpg', cv2.IMREAD_GRAYSCALE)

# Verificar si la imagen se cargó correctamente
if imagen is None:
    print("Error: No se pudo cargar la imagen. Verifica el nombre del archivo.")
else:
    # Obtener la profundidad de color original
    img_inicial = imagen.dtype
    print('Profundidad de color inicial:', img_inicial)  # Debería ser uint8 (8 bits)

    
    img_16bit = (imagen.astype(np.uint16)) * 256  # Multiplicamos por 256 para expandir el rango 0-255 → 0-65535


    img_nueva = img_16bit.dtype
    print('Profundidad de color nueva:', img_nueva)  # Debería ser uint16 (16 bits)

    # imagen or y conberrida
    plt.figure(figsize=(10, 5))

    # Imagen original (8 bits)
    plt.subplot(121)
    plt.title("Imagen Original (8 bits)")
    plt.imshow(imagen, cmap='gray')
    plt.axis("off")

    # Imagen convertida a 16 bits (Matplotlib no soporta 16 bits, pero se muestra como 8 bits para visualización)
    plt.subplot(122)
    plt.title("Imagen Convertida (16 bits)")
    plt.imshow(img_16bit // 256, cmap='gray')  # Normalizamos para que sea visible en 8 bits
    plt.axis("off")

    plt.tight_layout()
    plt.show()

    # Guardar la imagen convertida a 16 bits en el ordenador
    save_path = "imagen_16bit.png"
    cv2.imwrite(save_path, img_16bit)
    print(f"Imagen en 16 bits guardada en: {save_path}")
```

RGB A HSV

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

fig, axs = plt.subplots(1,4, figsize=(25,10))
img_bgr=cv2.imread('imagentango.jpg')
img_rgb=cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
hsv= cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)

axs[0].imshow(img_rgb)
axs[0].set_title('imagen')

fig_img1= axs[1].imshow(hsv[...,0],cmap='gray',vmin=0,vmax=179)
fig.colorbar(fig_img1, ax=axs[1], shrink=0.3)
axs[1].set_title('canal H')

fig_img2= axs[2].imshow(hsv[...,1],cmap='gray',vmin=0,vmax=255)
fig.colorbar(fig_img2, ax=axs[2], shrink=0.3)
axs[2].set_title('canal S')

fig_img3= axs[3].imshow(hsv[...,2],cmap='gray',vmin=0,vmax=255)
fig.colorbar(fig_img3, ax=axs[3], shrink=0.3)
axs[3].set_title('canal V')
```

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np

# Crear la figura con 4 subgráficos en una fila
fig, axs = plt.subplots(1, 4, figsize=(25, 10))

# Cargar la imagen en BGR y convertir a RGB
img_bgr = cv2.imread('imagentango.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Convertir la imagen de RGB a HSV
hsv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2HSV)

# Mostrar la imagen original en RGB
axs[0].imshow(img_rgb)
axs[0].set_title('Imagen Original')  # Corrección aquí
axs[0].axis("off")

# Mostrar el canal H (Tono) en escala de grises
fig_img1 = axs[1].imshow(hsv[..., 0], cmap='gray', vmin=0, vmax=179)
fig.colorbar(fig_img1, ax=axs[1], shrink=0.5)
axs[1].set_title('Canal H (Tono)')  # Corrección aquí
axs[1].axis("off")

# Mostrar el canal S (Saturación)
fig_img2 = axs[2].imshow(hsv[..., 1], cmap='gray', vmin=0, vmax=255)
fig.colorbar(fig_img2, ax=axs[2], shrink=0.5)
axs[2].set_title('Canal S (Saturación)')
axs[2].axis("off")

# Mostrar el canal V (Brillo)
fig_img3 = axs[3].imshow(hsv[..., 2], cmap='gray', vmin=0, vmax=255)
fig.colorbar(fig_img3, ax=axs[3], shrink=0.5)
axs[3].set_title('Canal V (Brillo)')
axs[3].axis("off")

# Ajustar el layout y mostrar
plt.tight_layout()
plt.show()
```

crear una imagen en rgb que contenga 4 cuadrantes en rojo, verde, azul y negro

```python
import matplotlib.pyplot as plt
import numpy as np

# definir tamano 200 x 200
tamano_imagen = 200  


imagen = np.zeros((tamano_imagen, tamano_imagen, 3), dtype=np.uint8)

# Definir la mitad de la imagen
mitad = tamano_imagen // 2

# dar colores a las partes
imagen[:mitad, :mitad] = [255, 0, 0]   
imagen[:mitad, mitad:] = [0, 255, 0]   
imagen[mitad:, :mitad] = [0, 0, 255]   
imagen[mitad:, mitad:] = [0, 0, 0]     

# Mostrar
plt.figure(figsize=(5, 5))
plt.imshow(imagen)
plt.axis("on") 
plt.title("Imagen con Cuadrantes de Colores")
plt.show()
```
