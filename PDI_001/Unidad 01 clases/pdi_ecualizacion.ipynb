ECUALIZACION DE UN HISTOGRAMA - SE DESEA NORMALIZAR UNA IMAGEN A UN PUNTO OPTIMO

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Leer la imagen
imagen = cv2.imread('imagentango.jpg', 0)

histograma,calc = np.histogram(imagen.flatten(),256, [0,256])

distAcum=histograma.cumsum()
distAcum_normalizado=distAcum+histograma.max()/distAcum.max()


ecualizado = cv2.equalizeHist(imagen)
histograma_equ,_ = np.histogram(ecualizado.flatten(), 256, [0,256])

plt.figure(figsize=(15,25))
plt.subplot(2,2,1)
plt.imshow(imagen, cmap='gray')

plt.subplot(2,2,2)
plt.plot(distAcum_normalizado, color='b')
plt.hist(imagen.flatten(), 256,[0,256], color='r')
plt.xlim([0,256])
plt.legend(('distribucion','Histograma'), loc='upper left')
plt.title('Histograma Original')


plt.subplot(2,2,3)
plt.imshow(ecualizado, cmap='gray')

plt.subplot(2,2,4)
plt.plot(histograma_equ, color='g')
plt.xlim([0,256])
plt.tight_layout
plt.show()
```

ecualizacion a colores

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10, 5]
imagen = cv2.imread('imagentango.jpg')
gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
ecualizada = cv2.equalizeHist(gray)
ecu_color = cv2.merge([
    cv2.equalizeHist(imagen[:, :, 0]), 
    cv2.equalizeHist(imagen[:, :, 1]), 
    cv2.equalizeHist(imagen[:, :, 2])   
])

plt.imshow(cv2.cvtColor(ecualizada, cv2.COLOR_BGR2RGB), cmap="gray")
plt.title("Equalized Grayscale Image")
plt.show()
plt.imshow(cv2.cvtColor(ecu_color, cv2.COLOR_BGR2RGB))
plt.title("Equalized Color Image")
plt.show()
```

ecualizacion sectorizada (CLAHE) POR CUADRANTES Y POR ESPACIOS

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Leer la imagen
imagen = cv2.imread('imagentango.jpg', 0)
clip=2.0
tiles=8

clahe = cv2.createCLAHE(clipLimit=clip, tileGridSize=(tiles, tiles))
ecualizado=clahe.apply(imagen)
plt.imshow(imagen, cmap='gray')
plt.show()
plt.imshow(ecualizado, cmap='gray')
plt.show()
```

ESPECTRO TRANSFORMADA DE FORIER

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Leer la imagen
imagen = cv2.imread('imagentango.jpg', 0)

transformada=np.fft.fft2(imagen)#aplico transformar de forier
desplazar = np.fft.fftshift(transformada) #separo el espectro
magnitud_espectro=np.log(1+np.abs(desplazar)) #calculo de espectro separado

plt.imshow(imagen, cmap='gray')
plt.show()
plt.imshow(magnitud_espectro, cmap='gray')
plt.show()


```
