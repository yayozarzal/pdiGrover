aplicacion de filtros bayer a mi imagen

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt


img_rgb = cv2.imread('imagentango.jpg')
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_BGR2GRAY)
bayer_pattern = np.zeros_like(img_gray)

# Simulación de patrón RGGB
bayer_pattern[0::2, 0::2] = img_gray[0::2, 0::2] 
bayer_pattern[0::2, 1::2] = img_gray[0::2, 1::2]  
bayer_pattern[1::2, 0::2] = img_gray[1::2, 0::2]  
bayer_pattern[1::2, 1::2] = img_gray[1::2, 1::2]  

# Aplicar distintos métodos de conversión Bayer -> BGR
demosaiced_bg = cv2.cvtColor(bayer_pattern, cv2.COLOR_BAYER_BG2BGR) 
demosaiced_gb = cv2.cvtColor(bayer_pattern, cv2.COLOR_BAYER_GB2BGR)  
demosaiced_rg = cv2.cvtColor(bayer_pattern, cv2.COLOR_BAYER_RG2BGR)  
demosaiced_gr = cv2.cvtColor(bayer_pattern, cv2.COLOR_BAYER_GR2BGR)  


plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(cv2.cvtColor(img_rgb, cv2.COLOR_BGR2RGB))
plt.title('Imagen Original')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(img_gray, cmap='gray')
plt.title('Escala de Grises')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(bayer_pattern, cmap='gray')
plt.title('Patrón Bayer Simulado')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(cv2.cvtColor(demosaiced_bg, cv2.COLOR_BGR2RGB))
plt.title('Demosaiced BGGR')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(cv2.cvtColor(demosaiced_gb, cv2.COLOR_BGR2RGB))
plt.title('Demosaiced GBRG')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(cv2.cvtColor(demosaiced_rg, cv2.COLOR_BGR2RGB))
plt.title('Demosaiced RGGB')
plt.axis('off')

plt.tight_layout()
plt.show()



```

transformacion yuv y sus canales

```python
import cv2
import matplotlib.pyplot as plt


img_bgr = cv2.imread('imagentango.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_yuv = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2YUV)


Y, U, V = cv2.split(img_yuv)
img_rgb_converted = cv2.cvtColor(img_yuv, cv2.COLOR_YUV2RGB)


plt.figure(figsize=(12, 6))

plt.subplot(2, 3, 1)
plt.imshow(img_rgb)
plt.title('Imagen Original (RGB)')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(Y, cmap='gray')
plt.title('Canal Y (Luminancia)')
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(U, cmap='gray')
plt.title('Canal U (Crominancia - Azul)')
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(V, cmap='gray')
plt.title('Canal V (Crominancia - Rojo)')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(img_rgb_converted)
plt.title('Convertido YUV → RGB')
plt.axis('off')

plt.tight_layout()
plt.show()
```
