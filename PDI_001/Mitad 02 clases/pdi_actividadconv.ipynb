ENFOQUE GRISES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
img=cv2.imread('imagentango.jpg',0)
mascara=np.array([[0,-1,0,],
                  [1,5,-1,],
                  [0,1,0,]])
img_final=cv2.filter2D(img,-1,mascara)
plt.subplot(121)
plt.title('Imagen Original')
plt.axis('off')
plt.imshow(img,cmap='gray')

plt.subplot(122)
plt.title('Imagen con mascara')
plt.axis('off')
plt.imshow(img_final,cmap='gray')

plt.show()
```

ENFOQUE COLORES

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('imagentango.jpg')


img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Definir la máscara
mascara = np.array([
    [0, -1, 0],
    [1, 5, -1],
    [0, 1, 0]
])


img_final = cv2.filter2D(img_rgb, -1, mascara)


plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.title('Imagen Original')
plt.axis('off')
plt.imshow(img_rgb)

plt.subplot(122)
plt.title('Imagen con máscara')
plt.axis('off')
plt.imshow(img_final)

plt.show()
```

FILTRO DE TIPO GAUSS

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('imagentango.jpg')


img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


mascara = np.array([
    [1, 2, 3, 1, 1],
    [2, 7, 11, 7, 2],
    [3, 11, 17, 11, 3],
    [2, 7, 11, 7, 1],
    [1, 2, 3, 2, 1]
]) / 115  


img_final = cv2.filter2D(img_rgb, -1, mascara)


plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.title('Imagen Original')
plt.axis('off')
plt.imshow(img_rgb)

plt.subplot(122)
plt.title('Imagen con filtro Gaussiano')
plt.axis('off')
plt.imshow(img_final)

plt.show()
```
