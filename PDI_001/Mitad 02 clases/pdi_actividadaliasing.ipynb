ALIASING Y PATRONES DE MOIRE


imagen 1

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


imagen = cv2.imread('imagen1.jpg', cv2.IMREAD_GRAYSCALE)

# 1. Suavizado Gaussiano
gauss = cv2.GaussianBlur(imagen, (9, 9), 2)

# 2. Suavizado Mediana
mediana = cv2.medianBlur(imagen, 5)

# 3. Filtro Bilateral
bilateral = cv2.bilateralFilter(imagen, d=9, sigmaColor=75, sigmaSpace=75)

# Mostrar comparaciones
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(imagen, cmap='gray')
plt.title("Imagen Original con Moiré")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(gauss, cmap='gray')
plt.title("Filtro Gaussiano")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(mediana, cmap='gray')
plt.title("Filtro de Mediana")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(bilateral, cmap='gray')
plt.title("Filtro Bilateral")
plt.axis("off")

plt.tight_layout()
plt.show()
```

aplicando convolucion diagonal

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


imagen = cv2.imread('imagen1.jpg', cv2.IMREAD_GRAYSCALE)

# Definir kernel
kernel = np.ones((5, 5), np.float32) / 25


imagen_convolucionada = cv2.filter2D(imagen, -1, kernel)

# Mostrar resultados
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(imagen, cmap='gray')
plt.title("Original con Moiré")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(imagen_convolucionada, cmap='gray')
plt.title("Suavizado con Convolución")
plt.axis("off")

plt.tight_layout()
plt.show()
```
