1. Genera una imagen a un tablero de ajedrez tomando en cuenta que cada casilla debe medir 100x100 px como se muestra en la imagen a continuacion (toma en cuenta la saturacion de los colores)

```python
import numpy as np
import matplotlib.pyplot as plt

# Definir tamaño del tablero
tamano_tablero = 10  # Número de casillas por lado
tamano_casilla = 100  # Tamaño de cada casilla en píxeles


tamano_imagen = tamano_tablero * tamano_casilla
tablero = np.zeros((tamano_imagen, tamano_imagen, 3), dtype=np.uint8)

# dibujar y dar colores
for fila in range(tamano_tablero):
    for columna in range(tamano_tablero):
        if (fila + columna) % 2 == 0:
            if(fila) % 2 ==1:
                tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                        columna * tamano_casilla:(columna + 1) * tamano_casilla] = [255, 255, 255]  # negro
            else:
               tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                        columna * tamano_casilla:(columna + 1) * tamano_casilla] = [0, 0, 0]  # blanco
            
            
        else:
            tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                    columna * tamano_casilla:(columna + 1) * tamano_casilla] = [100, 100, 100]  # Negro

# MOSTRAR EL TABLERO
plt.figure(figsize=(6, 6))
plt.imshow(tablero)
plt.axis("on") 
plt.title("EJERCICIO 1")
plt.show()
```

2. Genera una imagen como la de referencia

```python
import numpy as np
import matplotlib.pyplot as plt

# Definir tamaño del tablero
tamano_tablero = 7 # Número de casillas por lado
tamano_casilla = 100  # Tamaño de cada casilla en píxeles


tamano_imagen = tamano_tablero * tamano_casilla
tablero = np.zeros((tamano_imagen, tamano_imagen, 3), dtype=np.uint8)

# dibujar y dar colores
for fila in range(tamano_tablero):
    for columna in range(tamano_tablero):
        if fila == columna :
                tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                        columna * tamano_casilla:(columna + 1) * tamano_casilla] = [255, 255, 255]  # blanco
        elif ((fila+1) == columna) or (fila-6) == columna:
                tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                        columna * tamano_casilla:(columna + 1) * tamano_casilla] = [255, 255, 0]  # amaarillo
        elif ((fila+2) == columna) or (fila-5) == columna:
                tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                        columna * tamano_casilla:(columna + 1) * tamano_casilla] = [255, 127, 0]  # naranja
        elif ((fila+3) == columna) or (fila-4) == columna:
                tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                        columna * tamano_casilla:(columna + 1) * tamano_casilla] = [255, 0, 0]  # rojo
        elif ((fila+4) == columna) or (fila-3) == columna:
                tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                        columna * tamano_casilla:(columna + 1) * tamano_casilla] = [148, 0, 211]  # morado
        elif ((fila+5) == columna) or (fila-2) == columna:
                tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                        columna * tamano_casilla:(columna + 1) * tamano_casilla] = [0, 0, 255]  # azul
        elif ((fila+6) == columna) or (fila-1) == columna:
                tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                        columna * tamano_casilla:(columna + 1) * tamano_casilla] = [0, 255, 0] # verde
        else:
            tablero[fila * tamano_casilla:(fila + 1) * tamano_casilla,
                    columna * tamano_casilla:(columna + 1) * tamano_casilla] = [100, 100, 100]  # gris

# MOSTRAR EL TABLERO
plt.figure(figsize=(6, 6))
plt.imshow(tablero)
plt.axis("on") 
plt.title("EJERCICIO 2")
plt.show()
```

3. Haciendo uso de la imagen1.jpg obtiene el sigente resultado

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


imagen = cv2.imread('imagen1.jpg')
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)  


alto, ancho = imagen.shape[:2]


degradado = np.linspace(1, 0, alto).reshape(alto, 1, 1)  
degradado_rgb = np.repeat(degradado, ancho, axis=1)     
degradado_rgb = np.repeat(degradado_rgb, 3, axis=2)     


imagen_degradada = (imagen_rgb * degradado_rgb).astype(np.uint8)




plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(imagen_rgb)
plt.title("imagen original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(imagen_degradada)
plt.title("imagen con degradado lineal a colores")
plt.axis("off")

plt.tight_layout()
plt.show()
```

4. Haciendo uso de la imagen1.jpg obtiene el siguiente resultado

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


imagen = cv2.imread('imagen1.jpg')
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)


alto, ancho = imagen_rgb.shape[:2]


mascara = np.zeros((alto, ancho), dtype=np.uint8)


puntos_triangulo = np.array([[
    (ancho // 2, alto // 4),         
    (ancho // 4, 3 * alto // 4),     
    (3 * ancho // 4, 3 * alto // 4)  
]], dtype=np.int32)


cv2.fillPoly(mascara, puntos_triangulo, 255)


img_mascara = cv2.bitwise_and(imagen_rgb, imagen_rgb, mask=mascara)


plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(imagen_rgb)
plt.title('Imagen Original')
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(img_mascara)
plt.title('Imagen con Máscara Triangular')
plt.axis("off")

plt.tight_layout()
plt.show()
 

```

5. Tomando en cuenta la imagen1.jpg elige un punto de interes al azar (funcion randomica), genera su histograma por canales, solariza y posteriza la imagen, realiza un tratamiento de umbralizacion( el de preferencia), mezcla tecnicas y obtiene  un filtro artistico, lista todas las imagenes con sus correspondientes titulos, guarda todas las imagenes resultantes con sus nombres.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = [10, 5]
imagen = cv2.imread('imagen1.jpg')
imageng = cv2.imread('imagen1.jpg', 0)
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

# Umbral adaptativo
umbral_adap = cv2.adaptiveThreshold(imageng, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_cv2.THRESH_BINARY, 11, 2)
mask = umbral_adap == 255
img_junta = cv2.bitwise_and(imageng, imageng, mask=umbral_adap)

# \Funciones
def solarizar(imagen, umbral):
    return np.where(imagen < umbral, imagen, 255 - imagen)

def posterizar(imagen, nivel):
    factor = 255 / nivel
    return np.round(imagen / factor) * factor

def falso_color(imagen_gray):
    escalar = cv2.normalize(imagen_gray, None, 0, 255, cv2.NORM_MINMAX)
    return cv2.applyColorMap(escalar, cv2.COLORMAP_JET)

#Aplicar efectos
umbral = 127
imagen_solarizada = solarizar(imagen, umbral).astype(np.uint8)
nivel = 4
img_posterizada = posterizar(imagen, nivel).astype(np.uint8)
img_falso = falso_color(imagen)

#Punto de interes
alto, ancho = imagen_rgb.shape[:2]
y_random = np.random.randint(0, alto)
x_random = np.random.randint(0, ancho)
imagen_rgb_punto = imagen_rgb.copy()
cv2.circle(imagen_rgb_punto, (x_random, y_random), 6, (255, 0, 0), -1)  # punto azul

#Histograma por canales
color = ('r', 'g', 'b')
hist_data = [cv2.calcHist([imagen], [i], None, [256], [0, 256]) for i in range(3)]

# Mostrar resultados
plt.figure(figsize=(13, 8))

plt.subplot(2, 3, 1)
plt.imshow(imagen_rgb_punto)
plt.title(f'Imagen Original\n(Punto en [{x_random}, {y_random}])')
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(img_junta, cmap='gray')
plt.title('Umbral Adaptativo')
plt.axis('off')

plt.subplot(2, 3, 3)
for i, col in enumerate(color):
    plt.plot(hist_data[i], color=col)
plt.title('Histograma por Canales')
plt.xlabel("Intensidad")
plt.ylabel("Frecuencia")
plt.xlim([0, 256])

plt.subplot(2, 3, 4)
plt.imshow(cv2.cvtColor(imagen_solarizada, cv2.COLOR_BGR2RGB))
plt.title('Imagen Solarizada')
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(cv2.cvtColor(img_posterizada.astype(np.uint8), cv2.COLOR_BGR2RGB))
plt.title('Imagen Posterizada')
plt.axis('off')

plt.subplot(2, 3, 6)
plt.imshow(cv2.cvtColor(img_falso, cv2.COLOR_BGR2RGB))
plt.title('Filtro Artístico (JET)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
