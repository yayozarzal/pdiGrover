Identifica de cada uno de los elementos (elementos específicos) su área, perímetro y 
centroide, a cada uno de los objetos debe ser inscrito dentro un recuadro, una elipse, 
un círculo y marcar una línea divisora en el medio tomando de referencia el centroide 
(cada uno de estos conceptos en un subplot distinto) tomar en cuenta que el color de 
las figuras mencionadas debe ser el mismo que tiene el objeto seleccionado.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, Ellipse, Circle
import random

def generar_color_aleatorio():
    """Genera un color aleatorio para cada objeto"""
    return (random.randint(50, 255), random.randint(50, 255), random.randint(50, 255))

def clasificar_figura_por_vertices(contorno):
    """Clasifica una figura basándose en el número de vértices"""
    perimeter = cv2.arcLength(contorno, True)
    epsilon = 0.02 * perimeter
    approx = cv2.approxPolyDP(contorno, epsilon, True)
    +
    
    vertices = len(approx)
    
    if vertices == 3:
        return "Triangulo", vertices
    elif vertices == 4:
        x, y, w, h = cv2.boundingRect(contorno)
        aspect_ratio = float(w) / h
        if 0.9 <= aspect_ratio <= 1.1:
            return "Cuadrado", vertices
        else:
            return "Rectangulo", vertices
    elif vertices >= 8:
        return "Circulo", vertices
    else:
        return f"Poligono_{vertices}v", vertices

def calcular_elipse_inscrita(contorno):
    """Calcula los parámetros de la elipse que mejor se ajusta al contorno"""
    if len(contorno) >= 5:
        elipse = cv2.fitEllipse(contorno)
        return elipse
    return None

def calcular_circulo_inscrito(contorno):
    """Calcula el círculo inscrito en el contorno"""
    (x, y), radius = cv2.minEnclosingCircle(contorno)
    return (int(x), int(y)), int(radius)

def dibujar_linea_divisora(imagen, centroide, contorno, color):
    """Dibuja una línea divisora que pasa por el centroide"""
    # Calcular el rectángulo delimitador para determinar los límites
    x, y, w, h = cv2.boundingRect(contorno)
    
    # Línea horizontal que pasa por el centroide
    cv2.line(imagen, (x, centroide[1]), (x + w, centroide[1]), color, 2)
    
    # Línea vertical que pasa por el centroide
    cv2.line(imagen, (centroide[0], y), (centroide[0], y + h), color, 2)

def procesar_elemento_individual(imagen_original, contorno, color, elemento_id):
    """Procesa un elemento individual y crea los 4 subplots"""
    
    # Crear una máscara para aislar el elemento
    mask = np.zeros(imagen_original.shape[:2], dtype=np.uint8)
    cv2.drawContours(mask, [contorno], -1, 255, -1)
    
    # Extraer la región del elemento
    x, y, w, h = cv2.boundingRect(contorno)
    # Añadir margen para mejor visualización
    margen = 20
    x1, y1 = max(0, x - margen), max(0, y - margen)
    x2, y2 = min(imagen_original.shape[1], x + w + margen), min(imagen_original.shape[0], y + h + margen)
    
    elemento_recortado = imagen_original[y1:y2, x1:x2].copy()
    mask_recortado = mask[y1:y2, x1:x2]
    
    # Ajustar contorno a las nuevas coordenadas
    contorno_ajustado = contorno - np.array([x1, y1])
    
    # Calcular propiedades del elemento
    M = cv2.moments(contorno)
    if M['m00'] != 0:
        cx = int(M['m10'] / M['m00']) - x1
        cy = int(M['m01'] / M['m00']) - y1
    else:
        cx, cy = w // 2, h // 2
    
    centroide = (cx, cy)
    area = cv2.contourArea(contorno)
    perimetro = cv2.arcLength(contorno, True)
    
    # Clasificar la figura
    forma, vertices = clasificar_figura_por_vertices(contorno)
    
    # Crear las 4 versiones del elemento
    img_rectangulo = elemento_recortado.copy()
    img_elipse = elemento_recortado.copy()
    img_circulo = elemento_recortado.copy()
    img_linea_divisora = elemento_recortado.copy()
    
    # 1. Rectángulo delimitador
    rect_x, rect_y, rect_w, rect_h = cv2.boundingRect(contorno_ajustado)
    cv2.rectangle(img_rectangulo, (rect_x, rect_y), (rect_x + rect_w, rect_y + rect_h), color, 3)
    
    # 2. Elipse inscrita
    elipse = calcular_elipse_inscrita(contorno_ajustado)
    if elipse is not None:
        cv2.ellipse(img_elipse, elipse, color, 3)
    
    # 3. Círculo inscrito
    centro_circulo, radio = calcular_circulo_inscrito(contorno_ajustado)
    cv2.circle(img_circulo, centro_circulo, radio, color, 3)
    
    # 4. Línea divisora
    dibujar_linea_divisora(img_linea_divisora, centroide, contorno_ajustado, color)
    
    # Marcar centroide en todas las imágenes
    for img in [img_rectangulo, img_elipse, img_circulo, img_linea_divisora]:
        cv2.circle(img, centroide, 5, (255, 255, 255), -1)  # Blanco para contraste
        cv2.circle(img, centroide, 5, color, 2)  # Borde del color del objeto
    
    return {
        'id': elemento_id,
        'imagenes': {
            'rectangulo': img_rectangulo,
            'elipse': img_elipse,
            'circulo': img_circulo,
            'linea_divisora': img_linea_divisora
        },
        'propiedades': {
            'centroide': (cx + x1, cy + y1),  # Coordenadas originales
            'area': area,
            'perimetro': perimetro,
            'forma': forma,
            'vertices': vertices,
            'color': color,
            'bbox': (x, y, w, h)
        }
    }

def procesar_imagen_completa(ruta_imagen):
    """Procesa la imagen completa y analiza todos los elementos"""
    
    # Cargar imagen
    imagen = cv2.imread(ruta_imagen)
    if imagen is None:
        print(f"Error: No se pudo cargar la imagen {ruta_imagen}")
        return
    
    imagen_original = imagen.copy()
    gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
    gray = cv2.GaussianBlur(gray, (5, 5), 0)
    
    # Binarización
    ret, bin_img = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY_INV)
    
    # Encontrar contornos
    contornos, _ = cv2.findContours(bin_img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    # Filtrar y procesar elementos
    elementos_procesados = []
    colores_utilizados = []
    
    for i, contorno in enumerate(contornos):
        area = cv2.contourArea(contorno)
        if area < 500:  # Filtrar objetos muy pequeños
            continue
        
        # Generar color único para cada elemento
        color = generar_color_aleatorio()
        colores_utilizados.append(color)
        
        # Procesar elemento individual
        elemento = procesar_elemento_individual(imagen_original, contorno, color, i + 1)
        elementos_procesados.append(elemento)
    
    return elementos_procesados, imagen_original, bin_img

def mostrar_resultados_individuales(elementos):
    """Muestra los resultados de cada elemento en subplots separados"""
    
    if not elementos:
        print("No se encontraron elementos para procesar")
        return
    
    print("=== ANÁLISIS DETALLADO DE ELEMENTOS ===\n")
    
    for elemento in elementos:
        print(f"--- ELEMENTO {elemento['id']} ---")
        props = elemento['propiedades']
        print(f"Forma detectada: {props['forma']}")
        print(f"Área: {props['area']:.2f} píxeles²")
        print(f"Perímetro: {props['perimetro']:.2f} píxeles")
        print(f"Centroide: {props['centroide']}")
        print(f"Vértices: {props['vertices']}")
        print(f"Color RGB: {props['color']}")
        print()
        
        # Crear subplot para este elemento
        fig, axes = plt.subplots(2, 2, figsize=(12, 10))
        fig.suptitle(f'Elemento {elemento["id"]} - {props["forma"]} '
                    f'(Área: {props["area"]:.0f}px², Perímetro: {props["perimetro"]:.0f}px)', 
                    fontsize=14, fontweight='bold')
        
        # Subplot 1: Rectángulo delimitador
        axes[0, 0].imshow(cv2.cvtColor(elemento['imagenes']['rectangulo'], cv2.COLOR_BGR2RGB))
        axes[0, 0].set_title('Rectángulo Delimitador', fontweight='bold')
        axes[0, 0].axis('off')
        
        # Subplot 2: Elipse inscrita
        axes[0, 1].imshow(cv2.cvtColor(elemento['imagenes']['elipse'], cv2.COLOR_BGR2RGB))
        axes[0, 1].set_title('Elipse Inscrita', fontweight='bold')
        axes[0, 1].axis('off')
        
        # Subplot 3: Círculo inscrito
        axes[1, 0].imshow(cv2.cvtColor(elemento['imagenes']['circulo'], cv2.COLOR_BGR2RGB))
        axes[1, 0].set_title('Círculo Inscrito', fontweight='bold')
        axes[1, 0].axis('off')
        
        # Subplot 4: Línea divisora
        axes[1, 1].imshow(cv2.cvtColor(elemento['imagenes']['linea_divisora'], cv2.COLOR_BGR2RGB))
        axes[1, 1].set_title('Líneas Divisoras por Centroide', fontweight='bold')
        axes[1, 1].axis('off')
        
        plt.tight_layout()
        plt.show()

def mostrar_resumen_general(elementos, imagen_original, imagen_binaria):
    """Muestra un resumen general de todos los elementos detectados"""
    
    # Crear imagen con todos los elementos marcados
    imagen_resumen = imagen_original.copy()
    
    for elemento in elementos:
        props = elemento['propiedades']
        color = props['color']
        centroide = props['centroide']
        bbox = props['bbox']
        
        # Dibujar información en la imagen
        cv2.circle(imagen_resumen, centroide, 8, color, -1)
        cv2.rectangle(imagen_resumen, (bbox[0], bbox[1]), 
                     (bbox[0] + bbox[2], bbox[1] + bbox[3]), color, 3)
        
        # Etiqueta con información
        texto = f"E{elemento['id']}: {props['forma']}"
        cv2.putText(imagen_resumen, texto, (bbox[0], bbox[1] - 10), 
                   cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)
    
    # Mostrar resumen
    plt.figure(figsize=(15, 5))
    
    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(imagen_original, cv2.COLOR_BGR2RGB))
    plt.title('Imagen Original')
    plt.axis('off')
    
    plt.subplot(1, 3, 2)
    plt.imshow(imagen_binaria, cmap='gray')
    plt.title('Imagen Binaria')
    plt.axis('off')
    
    plt.subplot(1, 3, 3)
    plt.imshow(cv2.cvtColor(imagen_resumen, cv2.COLOR_BGR2RGB))
    plt.title(f'Elementos Detectados ({len(elementos)} elementos)')
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()
    
    # Mostrar estadísticas
    print(f"\n=== RESUMEN ESTADÍSTICO ===")
    print(f"Total de elementos detectados: {len(elementos)}")
    
    areas = [e['propiedades']['area'] for e in elementos]
    perimetros = [e['propiedades']['perimetro'] for e in elementos]
    
    print(f"Área promedio: {np.mean(areas):.2f} píxeles²")
    print(f"Área total: {np.sum(areas):.2f} píxeles²")
    print(f"Perímetro promedio: {np.mean(perimetros):.2f} píxeles")
    
    # Contar formas
    formas = [e['propiedades']['forma'] for e in elementos]
    from collections import Counter
    contador_formas = Counter(formas)
    
    print("\nDistribución de formas:")
    for forma, cantidad in contador_formas.items():
        print(f"  {forma}: {cantidad}")

# Función principal
def main(ruta_imagen):
    """Función principal que ejecuta todo el análisis"""
    
    try:
        print("Procesando imagen...")
        elementos, imagen_original, imagen_binaria = procesar_imagen_completa(ruta_imagen)
        
        if not elementos:
            print("No se detectaron elementos válidos en la imagen.")
            return
        
        print(f"Se detectaron {len(elementos)} elementos.")
        
        # Mostrar resumen general
        mostrar_resumen_general(elementos, imagen_original, imagen_binaria)
        
        # Mostrar análisis individual de cada elemento
        mostrar_resultados_individuales(elementos)
        
        # Conversión a centímetros (ajustar factor según necesidad)
        conversion_factor = 0.0264583333
        print(f"\n=== MEDIDAS EN CENTÍMETROS ===")
        for elemento in elementos:
            props = elemento['propiedades']
            area_cm2 = props['area'] * (conversion_factor ** 2)
            perimetro_cm = props['perimetro'] * conversion_factor
            print(f"Elemento {elemento['id']} ({props['forma']}): "
                  f"Área = {area_cm2:.2f} cm², Perímetro = {perimetro_cm:.2f} cm")
        
    except Exception as e:
        print(f"Error durante el procesamiento: {e}")

# Ejemplo de uso
if __name__ == "__main__":
    # Cambiar por la ruta de tu imagen
    ruta_imagen = 'img/colores.jpg'
    main(ruta_imagen)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import matplotlib.patches as patches

class ProcesadorImagenUnificado:
    def __init__(self, ruta_imagen):
        """Inicializa el procesador con la imagen especificada"""
        self.ruta_imagen = ruta_imagen
        self.imagen_original = None
        self.imagen_gray = None
        self.cargar_imagen()
    
    def cargar_imagen(self):
        """Carga la imagen original y la convierte a escala de grises"""
        self.imagen_original = cv2.imread(self.ruta_imagen)
        if self.imagen_original is None:
            raise ValueError(f"No se pudo cargar la imagen: {self.ruta_imagen}")
        self.imagen_gray = cv2.cvtColor(self.imagen_original, cv2.COLOR_BGR2GRAY)
        print(f"Imagen cargada exitosamente: {self.imagen_original.shape}")
    
    def aplicar_umbralizado_otsu(self):
        """Aplica umbralización usando el método de Otsu"""
        _, umbral_otsu = cv2.threshold(self.imagen_gray, 0, 255, 
                                      cv2.THRESH_BINARY + cv2.THRESH_OTSU)
        return umbral_otsu
    
    def aplicar_falso_color(self, mapa_color=cv2.COLORMAP_DEEPGREEN):
        """Aplica falso color a la imagen"""
        imagen_normalizada = cv2.normalize(self.imagen_gray, None, 0, 255, cv2.NORM_MINMAX)
        imagen_falso_color = cv2.applyColorMap(imagen_normalizada, mapa_color)
        return imagen_falso_color
    
    def aplicar_convolucion_personalizada(self, kernel_size=5):
        """Aplica convolución con kernel personalizado"""
        # Kernel de suavizado
        kernel = np.ones((kernel_size, kernel_size), np.float32) / (kernel_size ** 2)
        imagen_suavizada = cv2.filter2D(self.imagen_gray, -1, kernel)
        
        # Kernel personalizado (realce)
        kernel_realce = np.array([[0, -1, 0],
                                 [-1, 5, -1],
                                 [0, -1, 0]])
        imagen_realzada = cv2.filter2D(self.imagen_gray, -1, kernel_realce)
        
        return imagen_suavizada, imagen_realzada
    
    def aplicar_mascara_personalizada(self):
        """Aplica una máscara personalizada a la imagen"""
        imagen_rgb = cv2.cvtColor(self.imagen_original, cv2.COLOR_BGR2RGB)
        
        # Máscara personalizada
        mascara = np.array([[0, 1, 0],
                           [1, 4, 1],
                           [0, 1, 0]])
        
        imagen_con_mascara = cv2.filter2D(imagen_rgb, -1, mascara)
        return imagen_con_mascara
    
    def seleccionar_region_interes(self, x1, y1, x2, y2):
        """Selecciona una región de interés de la imagen"""
        imagen_rgb = cv2.cvtColor(self.imagen_original, cv2.COLOR_BGR2RGB)
        region_interes = imagen_rgb[y1:y2, x1:x2, :]
        return region_interes
    
    def detectar_bordes_sobel_mejorado(self, tam_kernel=3, umbral=30):
        """Detecta bordes usando Sobel mejorado"""
        # Suavizado gaussiano
        suavizado = cv2.GaussianBlur(self.imagen_gray, (3, 3), 0)
        
        # Gradientes Sobel
        sobel_x = cv2.Sobel(suavizado, cv2.CV_64F, 1, 0, ksize=tam_kernel)
        sobel_y = cv2.Sobel(suavizado, cv2.CV_64F, 0, 1, ksize=tam_kernel)
        
        # Magnitud
        magnitud = np.sqrt(sobel_x**2 + sobel_y**2)
        magnitud = cv2.normalize(magnitud, None, 0, 255, cv2.NORM_MINMAX, cv2.CV_8U)
        
        # Umbralización
        _, bordes = cv2.threshold(magnitud, umbral, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
        
        # Cierre morfológico
        kernel_morf = np.ones((2, 2), np.uint8)
        bordes = cv2.morphologyEx(bordes, cv2.MORPH_CLOSE, kernel_morf)
        
        return bordes
    
    def detectar_bordes_canny(self, low=100, high=200):
        """Detecta bordes usando Canny"""
        suavizado = cv2.GaussianBlur(self.imagen_gray, (3, 3), 0)
        return cv2.Canny(suavizado, low, high)
    
    def analizar_contornos_y_propiedades(self, imagen_binaria):
        """Analiza contornos y calcula propiedades geométricas"""
        contornos, _ = cv2.findContours(imagen_binaria, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        
        resultados = []
        imagen_con_analisis = cv2.cvtColor(self.imagen_original, cv2.COLOR_BGR2RGB).copy()
        
        for i, contorno in enumerate(contornos):
            area = cv2.contourArea(contorno)
            
            # Filtrar contornos muy pequeños
            if area < 500:
                continue
            
            # Calcular propiedades
            perimetro = cv2.arcLength(contorno, True)
            
            # Calcular momentos y centroide
            M = cv2.moments(contorno)
            if M['m00'] != 0:
                cx = int(M['m10'] / M['m00'])
                cy = int(M['m01'] / M['m00'])
            else:
                cx, cy = 0, 0
            
            # Bounding rectangle
            x, y, w, h = cv2.boundingRect(contorno)
            
            # Dibujar en la imagen
            # Contorno en verde
            cv2.drawContours(imagen_con_analisis, [contorno], -1, (0, 255, 0), 2)
            
            # Centroide en rojo
            cv2.circle(imagen_con_analisis, (cx, cy), 8, (255, 0, 0), -1)
            
            # Rectángulo delimitador en rosa
            cv2.rectangle(imagen_con_analisis, (x, y), (x + w, y + h), (255, 20, 147), 3)
            
            # Etiqueta con información
            texto = f"A:{area:.0f} P:{perimetro:.0f}"
            cv2.putText(imagen_con_analisis, texto, (x, y - 10), 
                       cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 20, 147), 2)
            
            resultados.append({
                'id': i + 1,
                'area': area,
                'perimetro': perimetro,
                'centroide': (cx, cy),
                'bbox': (x, y, w, h),
                'contorno': contorno
            })
        
        return resultados, imagen_con_analisis
    
    def procesar_imagen_completa(self):
        """Ejecuta todo el pipeline de procesamiento"""
        print("=== INICIANDO PROCESAMIENTO COMPLETO ===")
        
        # 1. Umbralización con Otsu
        print("1. Aplicando umbralización Otsu...")
        umbral_otsu = self.aplicar_umbralizado_otsu()
        
        # 2. Falso color
        print("2. Aplicando falso color...")
        falso_color = self.aplicar_falso_color()
        
        # 3. Convolución personalizada
        print("3. Aplicando convoluciones personalizadas...")
        suavizada, realzada = self.aplicar_convolucion_personalizada()
        
        # 4. Máscara personalizada
        print("4. Aplicando máscara personalizada...")
        con_mascara = self.aplicar_mascara_personalizada()
        
        # 5. Detección de bordes
        print("5. Detectando bordes...")
        bordes_sobel = self.detectar_bordes_sobel_mejorado()
        bordes_canny = self.detectar_bordes_canny()
        
        # 6. Análisis de contornos
        print("6. Analizando contornos y propiedades...")
        resultados_contornos, imagen_analizada = self.analizar_contornos_y_propiedades(umbral_otsu)
        
        # Mostrar todos los resultados
        self.mostrar_resultados_completos(
            umbral_otsu, falso_color, suavizada, realzada, con_mascara,
            bordes_sobel, bordes_canny, imagen_analizada, resultados_contornos
        )
        
        return resultados_contornos
    
    def mostrar_resultados_completos(self, umbral_otsu, falso_color, suavizada, realzada, 
                                   con_mascara, bordes_sobel, bordes_canny, imagen_analizada, 
                                   resultados_contornos):
        """Muestra todos los resultados en una visualización completa"""
        
        # Crear figura grande con múltiples subplots
        fig = plt.figure(figsize=(20, 15))
        
        # Fila 1: Imagen original y procesamiento básico
        plt.subplot(3, 4, 1)
        plt.imshow(cv2.cvtColor(self.imagen_original, cv2.COLOR_BGR2RGB))
        plt.title('1. Imagen Original', fontweight='bold')
        plt.axis('off')
        
        plt.subplot(3, 4, 2)
        plt.imshow(self.imagen_gray, cmap='gray')
        plt.title('2. Escala de Grises', fontweight='bold')
        plt.axis('off')
        
        plt.subplot(3, 4, 3)
        plt.imshow(umbral_otsu, cmap='gray')
        plt.title('3. Umbralización Otsu', fontweight='bold')
        plt.axis('off')
        
        plt.subplot(3, 4, 4)
        plt.imshow(cv2.cvtColor(falso_color, cv2.COLOR_BGR2RGB))
        plt.title('4. Falso Color', fontweight='bold')
        plt.axis('off')
        
        # Fila 2: Convoluciones y máscaras
        plt.subplot(3, 4, 5)
        plt.imshow(suavizada, cmap='gray')
        plt.title('5. Suavizado (Kernel 5x5)', fontweight='bold')
        plt.axis('off')
        
        plt.subplot(3, 4, 6)
        plt.imshow(realzada, cmap='gray')
        plt.title('6. Realzado (Kernel Personalizado)', fontweight='bold')
        plt.axis('off')
        
        plt.subplot(3, 4, 7)
        plt.imshow(con_mascara)
        plt.title('7. Máscara Personalizada', fontweight='bold')
        plt.axis('off')
        
        plt.subplot(3, 4, 8)
        plt.imshow(bordes_sobel, cmap='gray')
        plt.title('8. Bordes Sobel Mejorado', fontweight='bold')
        plt.axis('off')
        
        # Fila 3: Detección de bordes y análisis final
        plt.subplot(3, 4, 9)
        plt.imshow(bordes_canny, cmap='gray')
        plt.title('9. Bordes Canny', fontweight='bold')
        plt.axis('off')
        
        plt.subplot(3, 4, 10)
        plt.imshow(imagen_analizada)
        plt.title('10. Análisis de Contornos', fontweight='bold')
        plt.axis('off')
        
        # Gráfico de barras con áreas
        plt.subplot(3, 4, 11)
        if resultados_contornos:
            areas = [r['area'] for r in resultados_contornos]
            ids = [f"Obj {r['id']}" for r in resultados_contornos]
            plt.bar(ids, areas, color='pink')
            plt.title('11. Áreas de Objetos', fontweight='bold')
            plt.ylabel('Área (píxeles²)')
            plt.xticks(rotation=45)
        else:
            plt.text(0.5, 0.5, 'No se detectaron\nobjetos', ha='center', va='center', 
                    transform=plt.gca().transAxes, fontsize=12)
            plt.title('11. Áreas de Objetos', fontweight='bold')
        
        # Información de resumen
        plt.subplot(3, 4, 12)
        plt.axis('off')
        if resultados_contornos:
            resumen_texto = f"RESUMEN DE ANÁLISIS\\n\\n"
            resumen_texto += f"Objetos detectados: {len(resultados_contornos)}\\n\\n"
            
            for resultado in resultados_contornos:
                resumen_texto += f"Objeto {resultado['id']}:\\n"
                resumen_texto += f"  Área: {resultado['area']:.0f} px²\\n"
                resumen_texto += f"  Perímetro: {resultado['perimetro']:.0f} px\\n"
                resumen_texto += f"  Centroide: ({resultado['centroide'][0]}, {resultado['centroide'][1]})\\n\\n"
        else:
            resumen_texto = "No se detectaron objetos\\ncon área suficiente\\n(mínimo 500 píxeles²)"
        
        plt.text(0.1, 0.9, resumen_texto, transform=plt.gca().transAxes, 
                fontsize=10, verticalalignment='top', fontfamily='monospace')
        plt.title('12. Resumen de Resultados', fontweight='bold')
        
        plt.tight_layout()
        plt.show()
        
        # Imprimir resultados detallados
        self.imprimir_resultados_detallados(resultados_contornos)
    
    def imprimir_resultados_detallados(self, resultados):
        """Imprime los resultados detallados en consola"""
        print("\\n" + "="*50)
        print("RESULTADOS DETALLADOS DEL ANÁLISIS")
        print("="*50)
        
        if not resultados:
            print("No se detectaron objetos con área suficiente.")
            return
        
        print(f"Total de objetos detectados: {len(resultados)}\\n")
        
        for resultado in resultados:
            print(f"--- OBJETO {resultado['id']} ---")
            print(f"Área: {resultado['area']:.2f} píxeles²")
            print(f"Perímetro: {resultado['perimetro']:.2f} píxeles")
            print(f"Centroide: ({resultado['centroide'][0]}, {resultado['centroide'][1]})")
            print(f"Bounding Box (x, y, w, h): {resultado['bbox']}")
            print()
        
        # Estadísticas generales
        areas = [r['area'] for r in resultados]
        perimetros = [r['perimetro'] for r in resultados]
        
        print("--- ESTADÍSTICAS GENERALES ---")
        print(f"Área total: {sum(areas):.2f} píxeles²")
        print(f"Área promedio: {np.mean(areas):.2f} píxeles²")
        print(f"Perímetro promedio: {np.mean(perimetros):.2f} píxeles")
        print(f"Área máxima: {max(areas):.2f} píxeles²")
        print(f"Área mínima: {min(areas):.2f} píxeles²")
    
    def procesar_region_especifica(self, x1, y1, x2, y2):
        """Procesa una región específica de la imagen"""
        print(f"\\n=== PROCESANDO REGIÓN ESPECÍFICA ===")
        print(f"Coordenadas: ({x1}, {y1}) a ({x2}, {y2})")
        
        # Seleccionar región
        region = self.seleccionar_region_interes(x1, y1, x2, y2)
        
        # Aplicar filtros a la región
        region_gray = cv2.cvtColor(region, cv2.COLOR_RGB2GRAY)
        region_canny = cv2.Canny(region_gray, 50, 150)
        
        # Mostrar resultados de la región
        plt.figure(figsize=(15, 5))
        
        plt.subplot(1, 3, 1)
        plt.imshow(region)
        plt.title('Región Seleccionada')
        plt.axis('off')
        
        plt.subplot(1, 3, 2)
        plt.imshow(region_gray, cmap='gray')
        plt.title('Región en Escala de Grises')
        plt.axis('off')
        
        plt.subplot(1, 3, 3)
        plt.imshow(region_canny, cmap='gray')
        plt.title('Bordes Canny en Región')
        plt.axis('off')
        
        plt.tight_layout()
        plt.show()
        
        return region, region_canny

# Función principal para usar el procesador
def main():
    """Función principal que ejecuta todo el procesamiento"""
    
    
    ruta_imagen = 'img/retrato.jpg' 
    
    try:
        # Crear el procesador
        procesador = ProcesadorImagenUnificado(ruta_imagen)
        
        # Ejecutar procesamiento completo
        resultados = procesador.procesar_imagen_completa()
        
        # Opcional: Procesar una región específica
        # Ejemplo: procesar una región del ojo (ajustar coordenadas según tu imagen)
        # procesador.procesar_region_especifica(600, 650, 800, 850)
        
        print("\\n¡Procesamiento completado exitosamente!")
        
    except Exception as e:
        print(f"Error durante el procesamiento: {e}")
        print("Verifica que la ruta de la imagen sea correcta.")

# Ejecutar el programa
if __name__ == "__main__":
    main()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def detectar_objetos(imagen_path):
    
    imagen = cv2.imread(imagen_path)
    
    if imagen is None:
        print(f"Error: No se pudo cargar la imagen desde {imagen_path}. Verifica la ruta.")
        return None, None 

    imagen_original = imagen.copy() 
    

    gray = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

    gray = cv2.GaussianBlur(gray, (5, 5), 0)

    
    thresh = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                   cv2.THRESH_BINARY_INV, 11, 2)

   
    kernel = np.ones((3, 3), np.uint8)
    

    thresh = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel, iterations=2)
    
  
    thresh = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel, iterations=1)

    
    contornos, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    objetos_detectados = []
    imagen_resultado = imagen_original.copy()

    for i, contorno in enumerate(contornos):
        area = cv2.contourArea(contorno)
        
        # Filtrar contornos por un área mínima.
        # Ajusta este valor según el tamaño mínimo de los objetos que esperas detectar.
        if area < 500:  # descartar peques
            continue

        x, y, w, h = cv2.boundingRect(contorno)

    
        if w < 20 or h < 20: # descartar objetos muy delgados o pequeños
            continue

        # centro objeto
        centro_x = x + w // 2
        centro_y = y + h // 2

        # perimetro del contorno
        perimetro = cv2.arcLength(contorno, True)

    
        objeto = {
            'id': len(objetos_detectados) + 1, 
            'bbox': (x, y, w, h), 
            'centro': (centro_x, centro_y),
            'area': area,
            'perimetro': perimetro
        }
        objetos_detectados.append(objeto)

        
        cv2.rectangle(imagen_resultado, (x, y), (x + w, y + h), (0, 255, 0), 2) 

        
        cv2.circle(imagen_resultado, (centro_x, centro_y), 5, (255, 0, 0), -1) 

        
        cv2.putText(imagen_resultado, f"Objeto_{objeto['id']}",
                    (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 255, 0), 2) 

    
    plt.figure(figsize=(18, 6))

    plt.subplot(1, 3, 1) 
    plt.imshow(cv2.cvtColor(imagen_original, cv2.COLOR_BGR2RGB))
    plt.title('Imagen Original')
    plt.axis('off') 

    plt.subplot(1, 3, 2) 
    plt.imshow(thresh, cmap='gray') 
    plt.title('Imagen Procesada (Umbralizada)')
    plt.axis('off')

    plt.subplot(1, 3, 3) 
    plt.imshow(cv2.cvtColor(imagen_resultado, cv2.COLOR_BGR2RGB))
    plt.title(f'Objetos Detectados: {len(objetos_detectados)}')
    plt.axis('off')

    plt.tight_layout() 
    plt.show() 

    return objetos_detectados, imagen_resultado

def imprimir_resultados(objetos):
  
    print("\n--- INFORMACIÓN DE OBJETOS DETECTADOS ---\n")
    if not objetos:
        print("No se detectaron objetos para imprimir.")
        return

    for obj in objetos:
        print(f"--- OBJETO {obj['id']} ---")
        print(f"Posición (x, y, ancho, alto): {obj['bbox']}")
        print(f"Centro (x, y): {obj['centro']}")
        print(f"Área: {obj['area']:.2f} píxeles²") 
        print(f"Perímetro: {obj['perimetro']:.2f} píxeles") 
        print("-" * 20) 
    print("\n")


if __name__ == "__main__":
    
    ruta_imagen = 'img/colores.jpg'

    try:
        print(f"Iniciando la detección de objetos en: {ruta_imagen}...")
        
        objetos, imagen_resultado_final = detectar_objetos(ruta_imagen)

        if objetos:
            
            imprimir_resultados(objetos)
            
            nombre_archivo_salida = 'objetos_detectados.jpg'
            try:
                cv2.imwrite(nombre_archivo_salida, imagen_resultado_final)
                print(f"\n¡Detección completada! Se encontraron {len(objetos)} objetos.")
                print(f"Imagen con objetos detectados guardada como '{nombre_archivo_salida}'.")
            except Exception as e:
                print(f"Error al guardar la imagen '{nombre_archivo_salida}': {e}")
                print("Asegúrate de tener permisos de escritura en la carpeta.")
        else:
            print("No se detectaron objetos en la imagen proporcionada.")

    except Exception as e:
        # Capturar errores
        print(f"\nOcurrió un error inesperado durante la ejecución: {e}")
        print("Asegúrate de que OpenCV y Matplotlib estén instalados correctamente.")
        print("Puedes instalarlos con: pip install opencv-python matplotlib numpy")
```
