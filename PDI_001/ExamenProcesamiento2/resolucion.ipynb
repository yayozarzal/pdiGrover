```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

#conteo de elementos por color

image = cv2.imread('elementos.jpg')

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
_, binario=cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
borde = cv2.findContours(binario, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
cont = len(borde)

```
