creando una funcion, mostrar cada una de las imagenes en los dos modos de subplot ensenadors en los diferentes canales en escala de grises

```python
import cv2
import matplotlib.pyplot as plt

def mostrar_canales_varias(imagenes, modo=1):
    

    num_imgs = len(imagenes)  

    
    if num_imgs == 0:
        print("Error: No se proporcionaron imágenes.")
        return

  
    plt.figure(figsize=(num_imgs * 4, 6))

    for i, img_path in enumerate(imagenes):
       
        img_bgr = cv2.imread(img_path, cv2.IMREAD_COLOR)

        
        if img_bgr is None:
            print(f"Error: No se pudo cargar la imagen {img_path}. Verifica la ruta.")
            continue

        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

       
        canal_rojo = img_bgr[:, :, 2]  
        canal_verde = img_bgr[:, :, 1]  
        canal_azul = img_bgr[:, :, 0]  

        
        img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

        if modo == 1:
            
            plt.subplot(num_imgs, 4, i * 4 + 1)
            plt.title(f"Imagen {i+1} (RGB)")
            plt.imshow(img_rgb)
            plt.axis("off")

            plt.subplot(num_imgs, 4, i * 4 + 2)
            plt.title("Canal Rojo")
            plt.imshow(canal_rojo, cmap='gray')
            plt.axis("off")

            plt.subplot(num_imgs, 4, i * 4 + 3)
            plt.title("Canal Verde")
            plt.imshow(canal_verde, cmap='gray')
            plt.axis("off")

            plt.subplot(num_imgs, 4, i * 4 + 4)
            plt.title("Canal Azul")
            plt.imshow(canal_azul, cmap='gray')
            plt.axis("off")

    

        else:
            print("Modo no válido. Usa 1 para mostrar los canales o 2 para mostrar RGB y gris.")
            return

    plt.tight_layout()
    plt.show()



imagenes = ["img1.jpg", "img2.jpg", "img3.jpg"]


mostrar_canales_varias(imagenes, modo=1)


mostrar_canales_varias(imagenes, modo=2)
```

creando una funcion, mostrar cada una de las imagenes en los dos modos de subplot ensenadors en los diferentes canales de colore reales

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

def mostrar_canales_color(imagenes, modo=1):
    

    num_imgs = len(imagenes) 

    if num_imgs == 0:
        print("Error: No se proporcionaron imágenes.")
        return

    plt.figure(figsize=(num_imgs * 5, 6))

    for i, img_path in enumerate(imagenes):
        
        img_bgr = cv2.imread(img_path, cv2.IMREAD_COLOR)

      
        if img_bgr is None:
            print(f"Error: No se pudo cargar la imagen {img_path}. Verifica la ruta.")
            continue

        
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    
        r = img_bgr[:, :, 2]  
        g = img_bgr[:, :, 1]  
        b = img_bgr[:, :, 0]  

        
        red_imagen = np.zeros_like(img_rgb)
        red_imagen[:, :, 0] = r  

        green_imagen = np.zeros_like(img_rgb)
        green_imagen[:, :, 1] = g  

        blue_imagen = np.zeros_like(img_rgb)
        blue_imagen[:, :, 2] = b  

       
        img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

        if modo == 1:
          
            plt.subplot(num_imgs, 4, i * 4 + 1)
            plt.title(f"Imagen {i+1} (RGB)")
            plt.imshow(img_rgb)
            plt.axis("off")

            plt.subplot(num_imgs, 4, i * 4 + 2)
            plt.title("Canal Rojo")
            plt.imshow(red_imagen)
            plt.axis("off")

            plt.subplot(num_imgs, 4, i * 4 + 3)
            plt.title("Canal Verde")
            plt.imshow(green_imagen)
            plt.axis("off")

            plt.subplot(num_imgs, 4, i * 4 + 4)
            plt.title("Canal Azul")
            plt.imshow(blue_imagen)
            plt.axis("off")


        else:
            print("Modo no válido. Usa 1 para mostrar los canales o 2 para mostrar RGB y escala de grises.")
            return

    plt.tight_layout()
    plt.show()



imagenes = ["img1.jpg", "img2.jpg", "img3.jpg"]


mostrar_canales_color(imagenes, modo=1)


mostrar_canales_color(imagenes, modo=2)
```

```python
import cv2
import matplotlib.pyplot as plt

def mostrar_canales_falsos(imagenes):
    

    num_imgs = len(imagenes)  

    
    if num_imgs == 0:
        print("Error: No se proporcionaron imágenes.")
        return

    
    plt.figure(figsize=(num_imgs * 5, 6))

    for i, img_path in enumerate(imagenes):
        
        img_bgr = cv2.imread(img_path, cv2.IMREAD_COLOR)

        
        if img_bgr is None:
            print(f"Error: No se pudo cargar la imagen {img_path}. Verifica la ruta.")
            continue

        
        img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

        
        b, g, r = cv2.split(img_bgr)

        
        plt.subplot(num_imgs, 4, i * 4 + 1)
        plt.title(f"Imagen {i+1} (RGB)")
        plt.imshow(img_rgb)
        plt.axis("off")

        plt.subplot(num_imgs, 4, i * 4 + 2)
        plt.title("Canal Rojo (Falso)")
        plt.imshow(r, cmap="Reds")
        plt.axis("off")

        plt.subplot(num_imgs, 4, i * 4 + 3)
        plt.title("Canal Verde (Falso)")
        plt.imshow(g, cmap="Greens")
        plt.axis("off")

        plt.subplot(num_imgs, 4, i * 4 + 4)
        plt.title("Canal Azul (Falso)")
        plt.imshow(b, cmap="Blues")
        plt.axis("off")

    plt.tight_layout()
    plt.show()



imagenes = ["img1.jpg", "img2.jpg", "img3.jpg"]


mostrar_canales_falsos(imagenes)
```
